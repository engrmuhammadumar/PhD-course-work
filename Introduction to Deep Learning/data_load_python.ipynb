```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df_clean, x="species")
plt.title("Number of Penguins by Species")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df_clean["flipper_length_mm"], bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Flipper Length")
plt.xlabel("Flipper length (mm)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df_clean, x="bill_length_mm", y="bill_depth_mm", hue="species")
plt.title("Bill Length vs Bill Depth")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df_clean, x="species", y="body_mass_g")
plt.title("Body Mass by Species")
plt.show()
```

```python
avg_by_species = df_clean.groupby("species")["body_mass_g"].mean()

```

```python
plt.figure(figsize=(7, 4))
avg_by_species.plot(kind="bar", color=["#FF8C42", "#3498DB", "#2ECC71"])
plt.title("Average Body Mass by Species")
plt.ylabel("Body mass (g)")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
corr = df_clean.select_dtypes("number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Between Measurements")
plt.show()
```
